# Myocardial Infarction Complications Dataset

Este dataset, disponible en el [UCI Machine Learning Repository](https://archive.ics.uci.edu/dataset/579/myocardial+infarction+complications), contiene información clínica de 1700 pacientes hospitalizados por infarto agudo de miocardio (MI).

El objetivo principal es predecir la aparición de complicaciones y/o el resultado letal y su causa, utilizando información disponible en distintos momentos del periodo hospitalario.

## 📊 Estructura General

* Instancias: 1700 pacientes
* Features (columnas 2-112): 111 variables clínicas
* Targets (columnas 113-124): 12 variables de salida
* Valores faltantes: Sí

Las variables de entrada incluyen datos demográficos, signos clínicos, resultados de laboratorio, tratamientos administrados en UCI y otros indicadores médicos.

## ⏱ Escenarios Temporales de Predicción

El dataset permite realizar predicciones en cuatro momentos distintos del periodo hospitalario:

* **Al ingreso hospitalario:**<br> Se pueden usar todas las variables de entrada (2-112), excepto: 93, 94, 95, 100, 101, 102, 103, 104, 105

* **Al final del primer día (24h):**<br> Se pueden usar todas las variables excepto: 94, 95, 101, 102, 104, 105

* **Al final del segundo día (48h):**<br> Se pueden usar todas las variables excepto: 95, 102, 105

* **Al final del tercer día (72h):**<br> Se pueden usar todas las variables (2-112).


## 🎯 Problemas de Predicción

El dataset permite formular dos problemas principales de clasificación:

### 1️⃣ Predicción de Complicaciones (Multilabel)

Las siguientes 11 variables binarias indican si el paciente desarrolló una complicación durante la hospitalización:

<div align="center">

| # | Complicación |
|---|--------------|
| 1 | Fibrilación auricular |
| 2 | Taquicardia supraventricular |
| 3 | Taquicardia ventricular |
| 4 | Fibrilación ventricular |
| 5 | Bloqueo AV de tercer grado |
| 6 | Edema pulmonar |
| 7 | Ruptura miocárdica |
| 8 | Síndrome de Dressler |
| 9 | Insuficiencia cardíaca crónica |
| 10 | Reinfarto |
| 11 | Angina post-infarto |

</div>

Cada paciente puede presentar múltiples complicaciones simultáneamente, por lo que este problema se formula como: **Clasificación multilabel**

### 2️⃣ Mortalidad y Causa de Muerte (Multiclase)

La variable LET_IS indica el desenlace final:

<div align="center">

| Valor | Desenlace |
|-------|-----------|
| 0 | Vivo |
| 1 | Shock cardiogénico |
| 2 | Edema pulmonar |
| 3 | Ruptura miocárdica |
| 4 | Progresión de insuficiencia cardíaca |
| 5 | Tromboembolismo |
| 6 | Asistolia |
| 7 | Fibrilación ventricular |

</div>

Este problema se formula como: **Clasificación multiclase (8 clases)**

Es importante notar que algunas causas de muerte corresponden a complicaciones también modeladas como variables binarias; sin embargo, las variables binarias indican si ocurrió la complicación. LET_IS indica si esa complicación fue la causa final de muerte.

### 🧠 Relevancia Clínica

El infarto agudo de miocardio es una de las principales causas de mortalidad a nivel mundial. La evolución clínica es altamente variable: algunos pacientes no desarrollan complicaciones, mientras que otros presentan eventos graves que pueden conducir al fallecimiento.

Predecir oportunamente la aparición de complicaciones, el riesgo de muerte y su posible causa permite tomar medidas preventivas tempranas y mejorar el pronóstico del paciente.

```python
from ucimlrepo import fetch_ucirepo
from IPython.display import display, HTML
from scipy.stats import trim_mean
from numpy import mean, absolute

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
import math
```

```python
# obtener el dataset
myocardial_infarction_complications = fetch_ucirepo(id=579)

# datos (como dataframes de pandas)
X = myocardial_infarction_complications.data.features.copy()  # features
y_array = myocardial_infarction_complications.data.targets    # targets original

# convertir targets a dataframe
y = pd.DataFrame(y_array, columns=myocardial_infarction_complications.data.target_names)

# definir lista de variables numéricas y categóricas
numeric_cols = [
    "AGE", "S_AD_KBRIG", "D_AD_KBRIG", "S_AD_ORIT", "D_AD_ORIT",
    "K_BLOOD", "NA_BLOOD", "ALT_BLOOD", "AST_BLOOD", "KFK_BLOOD",
    "L_BLOOD", "ROE"
]
categorical_cols = [col for col in X.columns if col not in numeric_cols]

# corregir tipos de datos
X[numeric_cols] = X[numeric_cols].apply(lambda c: pd.to_numeric(c, errors='coerce'))
X[categorical_cols] = X[categorical_cols].apply(lambda c: pd.Categorical(c))
y = y.apply(lambda c: pd.Categorical(c))

# concatenar todo
data = pd.concat([X, y], axis=1)
data.head(10)
```

## **2. Definición de las Variables Objetivo: Complicaciones y Mortalidad**


```python
# ============================================================
# DEFINICIÓN DE VARIABLES OBJETIVO
# ============================================================

# ------------------------------------------------------------
# 1. Complicaciones médicas (Clasificación Multilabel)
# Cada variable indica si la complicación ocurrió (1) o no (0)
# ------------------------------------------------------------

comp_labels = {
    "FIBR_PREDS": "Fibrilación auricular",
    "PREDS_TAH": "Taquicardia supraventricular",
    "JELUD_TAH": "Taquicardia ventricular",
    "FIBR_JELUD": "Fibrilación ventricular",
    "A_V_BLOK": "Bloqueo AV grado III",
    "OTEK_LANC": "Edema pulmonar",
    "RAZRIV": "Ruptura miocárdica",
    "DRESSLER": "Síndrome de Dressler",
    "ZSN": "Falla cardiaca crónica",
    "REC_IM": "Reinfarto",
    "P_IM_STEN": "Angina post-infarto"
}

comp_cols = list(comp_labels.keys())
Y_comp = data[comp_cols]

# ------------------------------------------------------------
# 2. Mortalidad y causa de muerte (Clasificación Multiclase)
# 0 = Sobrevive, 1–7 = Causa específica de muerte
# ------------------------------------------------------------

letis_labels = {
    0: "Sobrevive",
    1: "Shock cardiogénico",
    2: "Edema pulmonar",
    3: "Ruptura miocárdica",
    4: "Falla cardiaca progresiva",
    5: "Tromboembolismo",
    6: "Asistolia",
    7: "Fibrilación ventricular"
}

Y_death = data["LET_IS"]
```

```python
# ============================================================
# INFORMACIÓN GENERAL DEL DATASET Y VARIABLES OBJETIVO
# ============================================================

print("Shape total:", data.shape)
print("Número de complicaciones:", len(comp_cols))
print("Número de clases de muerte:", Y_death.nunique())

pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)
```

```python
# ============================================================
# Análisis de frecuencia en las complicaciones médicas (enfoque multilabel)
# ============================================================

# Contar número de pacientes por complicación
comp_count = (Y_comp == 1).sum(axis=0).sort_values(ascending=False)
comp_percentage = (comp_count / len(Y_comp) * 100).loc[comp_count.index]

comp_names = comp_count.index.map(comp_labels)
comp_summary_df = pd.DataFrame({
    "Complication": comp_names,
    "Count": comp_count,
    "Percentage (%)": comp_percentage.map(lambda x: f"{x:.3f}%")
})

# ============================================================
# Figura: Tabla + Gráfica
# ============================================================
fig, axes = plt.subplots(1, 2, figsize=(14, 6), gridspec_kw={"width_ratios": [1.4, 1.2]})

# ---- TABLA ----
axes[0].axis("off")
table = axes[0].table(
    cellText=comp_summary_df.values,
    colLabels=comp_summary_df.columns,
    loc="center",
    cellLoc="center"
)

table.auto_set_font_size(False)
table.set_fontsize(10)
table.scale(0.9, 1.8)

for cell in table.get_celld().values():
    cell.set_text_props(ha="center", va="center")

axes[0].set_title("Complication Prevalence (Count & %)", pad=10)

# ---- GRÁFICA DE BARRAS ----
comp_percentage.plot(kind="bar", ax=axes[1])
axes[1].set_title("Complication Prevalence (%)")
axes[1].set_ylabel("Percentage of Patients")
axes[1].set_xlabel("Complication")
axes[1].tick_params(axis="x", rotation=45)

plt.tight_layout()
plt.show()
```

**Análisis**: El análisis de frecuencia de las complicaciones muestra una distribución claramente desbalanceada entre las distintas condiciones clínicas. La complicación más prevalente es la insuficiencia cardíaca crónica (23.18%), seguida por fibrilación auricular (10%), edema pulmonar (9.35%) y reinfarto (9.35%), lo que sugiere que estas condiciones representan los desenlaces adversos más comunes en pacientes hospitalizados por infarto de miocardio dentro del dataset. En contraste, complicaciones como taquicardia ventricular (2.47%) y taquicardia supraventricular (1.18%) aparecen con baja frecuencia, lo que evidencia un fuerte desbalance de clases que deberá considerarse en etapas posteriores de modelado. ## **3. Medidas de centralidad**

En esta sección se presentan las principales medidas de tendencia central y dispersión para las variables numéricas del dataset, con el fin de describir tanto el comportamiento típico como la variabilidad clínica de los pacientes hospitalizados por infarto agudo de miocardio.

**Media**: La media representa el valor promedio esperado de la variable y se obtiene como el promedio aritmético de todas las observaciones. Es sensible a valores extremos o atípicos, ya que cada valor influye directamente en el resultado.

$$
\bar{x} = \frac{1}{m} \sum_{i=1}^{m} x_i
$$

**Mediana**: La mediana es el valor central de los datos una vez ordenados. Divide el conjunto en dos partes iguales y es robusta frente a valores extremos, por lo que ofrece una medida más resistente del centro cuando la distribución es asimétrica.

$$
\text{Mediana} =
\begin{cases}
x_{(m+1)/2}, & \text{si } m \text{ es impar} \\
\frac{x_{(m/2)} + x_{(m/2+1)}}{2}, & \text{si } m \text{ es par}
\end{cases}
$$


**Moda**: La moda corresponde al valor que aparece con mayor frecuencia en el conjunto de datos. Es especialmente útil en variables discretas o categóricas, aunque también puede emplearse en variables numéricas.

$$
\text{Moda} = \arg\max_x f(x)
$$

donde \( f(x) \) representa la frecuencia del valor \( x \).

```python
# ============================================================
# Medias de centralidad (numéricas)
# ============================================================

def summarize_numeric(df, numeric_cols):
    summary = pd.DataFrame({
        "Min": df[numeric_cols].min(),
        "Max": df[numeric_cols].max(),
        "Mean": df[numeric_cols].mean().round(2),
        "Median": df[numeric_cols].median().round(2),
        "Missing": df[numeric_cols].isna().sum()
    })
    summary["Missing %"] = (summary["Missing"] / len(df) * 100).round(2).astype(str) + "%"
    return summary

numeric_summary = summarize_numeric(data, numeric_cols)
print("===================== Resumen Variables Numéricas =====================")
display(numeric_summary)
```

## **4. Medidas de dispersión**

**Rango**: El rango mide la amplitud total de los datos y permite identificar la extensión completa de los valores observados, siendo sensible a valores extremos.

$$
\text{Rango} = x_{\max} - x_{\min}
$$

**Varianza**: La varianza cuantifica la dispersión promedio cuadrática respecto a la media. Se utiliza la versión muestral (con corrección de Bessel):

$$
\text{Var}(x) = \frac{1}{m-1} \sum_{i=1}^{m} (x_i - \bar{x})^2
$$

**Desviación Estándar**: La desviación estándar es la raíz cuadrada de la varianza. Su principal ventaja es que se expresa en las mismas unidades que la variable original, facilitando la interpretación clínica.

$$
\text{SD}(x) = \sqrt{\text{Var}(x)}
$$

**Desviación Media Absoluta (MAD)**: La Desviación Media Absoluta mide la dispersión promedio en términos absolutos respecto a una medida central. En este análisis se utilizó la mediana como referencia, lo que la hace más robusta frente a valores extremos.

$$
\text{MAD}(x) = \frac{1}{m} \sum_{i=1}^{m} |x_i - \tilde{x}|
$$

donde \( $ \tilde{x} $ \) representa la mediana.

```python
# ============================================================
# Medidas de dispersión
# ============================================================

# función para Average Absolute Deviation (AAD)
def aad(x, center=np.median):
    x = np.array(x.dropna())  # eliminar NaN
    return np.mean(np.abs(x - center(x)))

# crear DataFrame vacío
resumen_dispersion = pd.DataFrame(columns=['Rango', 'Varianza', 'Desv_Estandar', 'AAD'])

# calcular medidas para cada variable numérica
for col in numeric_cols:
    resumen_dispersion.loc[col] = [
        data[col].max() - data[col].min(),  # Rango
        data[col].var(),                    # Varianza
        data[col].std(),                    # Desviación estándar
        aad(data[col])                      # AAD
    ]

resumen_dispersion
```

* **Edad de los pacientes**: La variable AGE presenta un rango de 66 años, con una desviación estándar de 11.26 y una AAD de 8.91. Esto indica una dispersión moderada en la edad de los pacientes, lo que sugiere que el dataset incluye individuos de distintos grupos etarios, aunque concentrados alrededor de la media.

**Covarianza**: La covarianza mide el grado en que dos variables numéricas varían conjuntamente. Permite identificar si existe una relación lineal entre ellas y la dirección de dicha relación.

- Si la covarianza es positiva, ambas variables tienden a aumentar o disminuir conjuntamente.
- Si es negativa, cuando una variable aumenta, la otra tiende a disminuir.
- Si es cercana a cero, no existe una relación lineal apreciable.

Se utiliza la versión muestral (con corrección de Bessel):

$$
\text{Cov}(X, Y) =
\frac{1}{m-1}
\sum_{i=1}^{m}
(x_i - \bar{x})(y_i - \bar{y})
$$

donde \( $ \bar{x} $ \) y \( $ \bar{y} $ \) representan las medias muestrales de las variables \( X \) y \( Y \), respectivamente.

La covarianza depende de las unidades de medida de las variables, por lo que su magnitud no es directamente interpretable sin estandarización.

```python
# ============================================================
# Matriz de Correlación (Covarianza Estandarizada)
# ============================================================

# Usamos .corr() en lugar de .cov() para obtener valores entre -1 y 1
corr_matrix = data[numeric_cols].corr()

plt.figure(figsize=(10, 8))

# Agregamos vmin y vmax para que la escala de colores siempre sea de -1 a 1
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap="coolwarm", vmin=-1, vmax=1)

plt.title("Matriz de Correlación (Interpretable)")
plt.show()
```

## **5. Percentiles, tablas de frecuencia y tortas**

En esta sección se presentan herramientas adicionales para resumir y visualizar las variables del dataset, con el fin de analizar su distribución, frecuencia y comportamiento estructural dentro de la población de pacientes hospitalizados por infarto agudo de miocardio.

**Percentiles**: Los percentiles dividen el conjunto de datos ordenado en 100 partes iguales y permiten determinar la posición relativa de una observación dentro de la distribución. Indican el valor por debajo del cual se encuentra un determinado porcentaje de los datos.

$$
P_k = \text{valor que deja } k\% \text{ de los datos por debajo}
$$

Por ejemplo, el percentil 25 (primer cuartil, Q1) indica que el 25% de los pacientes presentan valores menores o iguales a ese punto. Los percentiles más utilizados en análisis descriptivo son 0%, 10%, 25%, 50%, 75%, 90% y 100%, ya que permiten caracterizar la dispersión y la asimetría de la distribución.

```python
# ============================================================
# Percentiles de variables numéricas para todo el dataset
# ============================================================

def numeric_percentiles(df, numeric_cols, percentiles=[0, 0.10, 0.25, 0.50, 0.75, 0.90, 1.0]):
    summary = df[numeric_cols].quantile(percentiles).T
    summary.columns = [f"{int(p*100)}%" for p in percentiles]
    summary["Missing"] = df[numeric_cols].isna().sum()
    summary["Missing %"] = (summary["Missing"] / len(df) * 100).map(lambda x: f"{x:.2f}%")
    return summary

numeric_quantiles = numeric_percentiles(data, numeric_cols)
display(numeric_quantiles)
```

```python
# ============================================================
# Percentiles de variables numéricas por complicación médica
# ============================================================

comp_numeric_quantiles = {}
for comp in comp_cols:
    comp_data = data[data[comp] == 1]
    comp_numeric_quantiles[comp_labels[comp]] = numeric_percentiles(comp_data, numeric_cols)

# ---- Mortalidad global (LET_IS != 0) ----
death_data = data[data["LET_IS"] != 0]
comp_numeric_quantiles["Mortalidad"] = numeric_percentiles(death_data, numeric_cols)

tabla_comp_numeric_quantiles = pd.concat(comp_numeric_quantiles, names=["Complicacion", "Variable_numerica"])
display(tabla_comp_numeric_quantiles)
```

$$
\text{Proporción} =
\frac{\text{número de casos de la categoría}}
{\text{total de casos}}
\times 100\%
$$

Este tipo de representación permite comparar visualmente la participación relativa de cada categoría y detectar aquellas predominantes dentro de la muestra.

```python
# ============================================================
# Graficar diagramas de torta para variables categóricas
# ============================================================

# Número de variables categóricas
n_vars = len(categorical_cols)

# Definir grilla: por ejemplo 3 columnas y filas necesarias
n_cols = 3
n_rows = (n_vars // n_cols) + (n_vars % n_cols > 0)

fig, axes = plt.subplots(n_rows, n_cols, figsize=(18, n_rows*5))
axes = axes.flatten()  # para iterar fácilmente

for i, col in enumerate(categorical_cols):
    # Conteo de cada categoría, ignorando NaN
    counts = data[col].value_counts(dropna=False)

    # Gráfico de torta
    wedges, texts, autotexts = axes[i].pie(
        counts,
        labels=counts.index,
        autopct='%1.1f%%', startangle=90
    )

    axes[i].set_title(col)
    axes[i].axis('equal')  # hace el gráfico circular

    axes[i].legend(wedges, [f"{val}" for val in counts.index],
                   title="Categorías", loc="center left", bbox_to_anchor=(1, 0, 0.5, 1))

# Eliminar ejes vacíos si sobran subplots
for j in range(len(categorical_cols), len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

A continuación, se analiza la relación entre el sexo y los distintos desenlaces clínicos, explorando si existen patrones diferenciales de mortalidad o supervivencia entre hombres y mujeres.
```python
# ============================================================
# Grupo etario automático cada 10 años
# ============================================================

age_min = (data["AGE"].min() // 10) * 10
age_max = ((data["AGE"].max() // 10) + 1) * 10

age_bins = range(int(age_min), int(age_max) + 10, 10)
data["Grupo_edad"] = pd.cut(data["AGE"], bins=age_bins, right=False)

letis_palette = {
    "Sobrevive": "#2ca02c",
    "Shock cardiogénico": "#d62728",
    "Edema pulmonar": "#ff7f0e",
    "Ruptura miocárdica": "#1f77b4",
    "Falla cardiaca progresiva": "#9467bd",
    "Tromboembolismo": "#8c564b",
    "Asistolia": "#e377c2",
    "Fibrilación ventricular": "#7f7f7f"
}

sex_palette = {
    'Hombre': "#1f77b4",
    'Mujer': "#ff7f0e",
    "Missing": "#cccccc"
}

data["LET_IS_label"] = data["LET_IS"].map(letis_labels)
data_fallecidos = data[data["LET_IS"] != 0].copy()

tabla_comp_norm = (
    data_fallecidos
    .groupby(["num_complications", "LET_IS_label"], observed=True)
    .size()
    .reset_index(name="n")
)

tabla_comp_norm["porcentaje"] = (
    tabla_comp_norm
    .groupby("num_complications", observed=True)["n"]
    .transform(lambda x: x / x.sum() * 100)
)

tabla_edad_comp = (
    data_fallecidos
    .groupby(
        ["Grupo_edad", "num_complications", "LET_IS_label"],
        observed=True
    )
    .size()
    .reset_index(name="n")
)

tabla_edad_comp["porcentaje"] = (
    tabla_edad_comp
    .groupby(["Grupo_edad", "num_complications"], observed=True)["n"]
    .transform(lambda x: x / x.sum() * 100)
)

tabla_edad_comp["Edad_num"] = tabla_edad_comp["Grupo_edad"].apply(lambda x: x.mid)
```

### Figura 3. Complicaciones cardiovasculares agudas según desenlace clínico

Esta figura examina la relación entre complicaciones cardiovasculares agudas —como fibrilación ventricular, bloqueo auriculoventricular, edema pulmonar y ruptura cardíaca— y el desenlace clínico de los pacientes. La visualización de la frecuencia de estas complicaciones estratificada por desenlace permite identificar cuáles eventos se asocian con mayor mortalidad o con desenlaces clínicos más graves. De este modo, la figura contribuye a reconocer complicaciones críticas que podrían tener un papel determinante en la evolución clínica posterior al infarto.

```python
# ============================================================
# Figura 3 - Complicaciones cardiovasculares agudas y desenlace
# ============================================================

sns.set_style("whitegrid")

# Excluir pacientes vivos
data_plot = data[data["LET_IS"] != 0]
comp_counts = pd.DataFrame()

for comp_col, comp_name in comp_labels.items():
    df_sub = data_plot[data_plot[comp_col] == 1].copy()
    counts = df_sub["LET_IS_label"].value_counts().reset_index()
    counts.columns = ["Desenlace", "Pacientes"]
    counts["Complicación"] = comp_name
    comp_counts = pd.concat([comp_counts, counts], axis=0)

# Separar en dos grupos para graficar
comp_keys = list(comp_labels.keys())
group1 = comp_keys[:6]
group2 = comp_keys[6:]

fig, axes = plt.subplots(2, 1, figsize=(16, 10), sharey=True)

for i, group in enumerate([group1, group2]):
    # Obtener nombres legibles para este grupo
    group_names = [comp_labels[c] for c in group]

    sns.barplot(
        data=comp_counts[comp_counts["Complicación"].isin(group_names)],
        x="Complicación",
        y="Pacientes",
        hue="Desenlace",
        palette=letis_palette,
        ax=axes[i]
    )
    axes[i].set_ylabel("Número de pacientes")
    axes[i].set_xlabel("")
    axes[i].tick_params(axis='x', rotation=45)
    axes[i].set_title(f"Complicaciones cardiovasculares (grupo {i+1})")
    axes[i].get_legend().remove()  # solo leyenda externa

# Leyenda global
handles, labels = axes[0].get_legend_handles_labels()
fig.legend(
    handles,
    labels,
    title="Causa de muerte",
    loc="center left",
    bbox_to_anchor=(0.92, 0.5),
    borderaxespad=0
)

plt.tight_layout(rect=[0, 0, 0.90, 1])
plt.show()
```

**Análisis**

La visualización permite comparar con mayor facilidad la frecuencia de cada causa de muerte asociada a las distintas complicaciones cardiovasculares agudas.

En términos generales, los patrones observados coinciden con los identificados en el análisis tabular inicial. Entre ellos destaca particularmente el caso de la ruptura miocárdica, en la cual se observa que la totalidad de los pacientes afectados fallece a causa de esta complicación, lo que evidencia la alta gravedad clínica asociada a este evento.

### Figura 4. Asociación entre tratamientos, eventos recurrentes y desenlace clínico

Esta figura analiza la relación entre distintos tratamientos farmacológicos y la aparición de eventos recurrentes o complicaciones posteriores, considerando su impacto en el desenlace clínico. La normalización de los resultados respecto al total de pacientes tratados o no tratados permite comparar de forma proporcional la mortalidad y la supervivencia asociadas a cada intervención. Este análisis facilita la evaluación exploratoria de la efectividad relativa de los tratamientos y su posible influencia en la evolución clínica de los pacientes tras el infarto.

```python
# ============================================================
# Figura 4 revisada: Asociación entre tratamientos y desenlace (LET_IS != 0)
# ============================================================

tratamientos = [
    ("fibr_ter_01", "Celiasum 750k UI"),
    ("fibr_ter_02", "Celiasum 1M UI"),
    ("fibr_ter_03", "Celiasum 3M UI"),
    ("fibr_ter_05", "Streptase"),
    ("B_BLOK_S_n", "Betabloqueadores"),
    ("ASP_S_n", "Ácido acetilsalicílico"),
    ("GEPAR_S_n", "Heparina"),
    ("LID_S_n", "Lidocaína")
]

# Filtrar solo pacientes con desenlace adverso
data_fallecidos = data[data["LET_IS"] != 10].copy()

def tabla_normalizada(df, var):
    tabla = (
        df
        .groupby([var, "LET_IS_label"], observed=True)
        .size()
        .reset_index(name="n")
    )
    tabla["Porcentaje"] = tabla.groupby(var, observed=True)["n"].transform(lambda x: 100 * x / x.sum())
    return tabla

fig, axes = plt.subplots(2, 4, figsize=(22, 10))
sns.set_style("whitegrid")

for i, (var, title) in enumerate(tratamientos):
    row, col = divmod(i, 4)
    df_norm = tabla_normalizada(data_fallecidos, var)

    sns.barplot(
        data=df_norm,
        x=var,
        y="Porcentaje",
        hue="LET_IS_label",
        palette=letis_palette,
        ax=axes[row, col]
    )

    axes[row, col].set_title(title, fontsize=12)
    axes[row, col].set_xlabel("Tratamiento (0 = No, 1 = Sí)")
    axes[row, col].set_ylabel("% de pacientes")
    axes[row, col].legend_.remove()  # quitamos leyenda interna

# Leyenda global
handles = [plt.Rectangle((0,0),1,1, color=c) for c in letis_palette.values()]
fig.legend(handles, letis_palette.keys(), title="Desenlace clínico",
           loc="center left", bbox_to_anchor=(0.92, 0.5))

plt.suptitle("Asociación entre tratamientos y desenlace clínico (pacientes fallecidos)", fontsize=20, y=1.02)
plt.tight_layout(rect=[0,0,0.90,0.95])
plt.show()
```

**Análisis**:

La figura muestra la relación entre distintos tratamientos farmacológicos y el desenlace clínico en los pacientes analizados.

* En el caso de Celiasum 750k UI, se observa que el porcentaje de pacientes que sobrevive es ligeramente mayor entre quienes recibieron el tratamiento.

* Para Celiasum 1M UI, el porcentaje de supervivencia disminuye aproximadamente un 15 % en los pacientes que lo recibieron.

* En Celiasum 3M UI, también se aprecia una leve reducción en el número de pacientes que sobreviven entre quienes recibieron el tratamiento, aunque la diferencia es pequeña.

* En contraste, el tratamiento con Streptase muestra un aumento notable en la proporción de pacientes que sobreviven, alcanzando valores cercanos al total de los casos tratados.

* Respecto a otros tratamientos, los betabloqueadores y el ácido acetilsalicílico también presentan un incremento en la proporción de supervivientes entre los pacientes que los recibieron.

* En el caso de la heparina, se observa un aumento leve en la supervivencia asociado a su uso.

* Finalmente, para la lidocaína, la proporción de supervivientes se mantiene prácticamente igual, sin diferencias claras entre pacientes tratados y no tratados.
