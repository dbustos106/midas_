# Myocardial Infarction Complications Dataset

Este dataset, disponible en el [UCI Machine Learning Repository](https://archive.ics.uci.edu/dataset/579/myocardial+infarction+complications), contiene información clínica de 1700 pacientes hospitalizados por infarto agudo de miocardio (MI).

El objetivo principal es predecir la aparición de complicaciones y/o el resultado letal y su causa, utilizando información disponible en distintos momentos del periodo hospitalario.

## 📊 Estructura General

* Instancias: 1700 pacientes
* Features (columnas 2-112): 111 variables clínicas
* Targets (columnas 113-124): 12 variables de salida
* Valores faltantes: Sí

Las variables de entrada incluyen datos demográficos, signos clínicos, resultados de laboratorio, tratamientos administrados en UCI y otros indicadores médicos.

## ⏱ Escenarios Temporales de Predicción

El dataset permite realizar predicciones en cuatro momentos distintos del periodo hospitalario:

* **Al ingreso hospitalario:**<br> Se pueden usar todas las variables de entrada (2-112), excepto: 93, 94, 95, 100, 101, 102, 103, 104, 105

* **Al final del primer día (24h):**<br> Se pueden usar todas las variables excepto: 94, 95, 101, 102, 104, 105

* **Al final del segundo día (48h):**<br> Se pueden usar todas las variables excepto: 95, 102, 105

* **Al final del tercer día (72h):**<br> Se pueden usar todas las variables (2-112).


## 🎯 Problemas de Predicción

El dataset permite formular dos problemas principales de clasificación:

### 1️⃣ Predicción de Complicaciones (Multilabel)

Las siguientes 11 variables binarias indican si el paciente desarrolló una complicación durante la hospitalización:

<div align="center">

| # | Complicación |
|---|--------------|
| 1 | Fibrilación auricular |
| 2 | Taquicardia supraventricular |
| 3 | Taquicardia ventricular |
| 4 | Fibrilación ventricular |
| 5 | Bloqueo AV de tercer grado |
| 6 | Edema pulmonar |
| 7 | Ruptura miocárdica |
| 8 | Síndrome de Dressler |
| 9 | Insuficiencia cardíaca crónica |
| 10 | Reinfarto |
| 11 | Angina post-infarto |

</div>

Cada paciente puede presentar múltiples complicaciones simultáneamente, por lo que este problema se formula como: **Clasificación multilabel**

### 2️⃣ Mortalidad y Causa de Muerte (Multiclase)

La variable LET_IS indica el desenlace final:

<div align="center">

| Valor | Desenlace |
|-------|-----------|
| 0 | Vivo |
| 1 | Shock cardiogénico |
| 2 | Edema pulmonar |
| 3 | Ruptura miocárdica |
| 4 | Progresión de insuficiencia cardíaca |
| 5 | Tromboembolismo |
| 6 | Asistolia |
| 7 | Fibrilación ventricular |

</div>

Este problema se formula como: **Clasificación multiclase (8 clases)**

Es importante notar que algunas causas de muerte corresponden a complicaciones también modeladas como variables binarias; sin embargo, las variables binarias indican si ocurrió la complicación. LET_IS indica si esa complicación fue la causa final de muerte.

### 🧠 Relevancia Clínica

El infarto agudo de miocardio es una de las principales causas de mortalidad a nivel mundial. La evolución clínica es altamente variable: algunos pacientes no desarrollan complicaciones, mientras que otros presentan eventos graves que pueden conducir al fallecimiento.

Predecir oportunamente la aparición de complicaciones, el riesgo de muerte y su posible causa permite tomar medidas preventivas tempranas y mejorar el pronóstico del paciente.

## 3. **Imputación de datos faltantes**

Se implementa una estrategia diferenciada según el tipo y porcentaje de datos faltantes. Para variables categóricas, los valores NaN se recodifican como "Missing". En las numéricas, se aplica imputación según el grado de ausencia: se reemplaza por mediana cuando el missing es bajo (<15%), por media cuando es moderado (15-30%) y se crea una columna indicadora _missing para preservar la información de ausencia. Las variables con más del 30% de datos faltantes se eliminan.

```python
# ============================================================
# Manejo de datos faltantes
# ============================================================

# ------------------------------------------------------------
# 1. Funciones de summarize
# ------------------------------------------------------------

def summarize_categorical(df, categorical_cols):
    summary = pd.DataFrame({
        "N_Cases": df[categorical_cols].nunique(),
        "Mode": df[categorical_cols].mode().iloc[0],
        "Missing": df[categorical_cols].isna().sum()
    })
    summary["Missing %"] = (summary["Missing"] / len(df) * 100).round(2).astype(str) + "%"
    return summary

def summarize_numeric(df, numeric_cols):
    s = pd.DataFrame({
        "Min": df[numeric_cols].min(),
        "Max": df[numeric_cols].max(),
        "Mean": df[numeric_cols].mean().round(2),
        "Median": df[numeric_cols].median().round(2),
        "Missing": df[numeric_cols].isna().sum()
    })
    s["Missing %"] = (s["Missing"] / len(df) * 100).round(2).astype(str) + "%"
    return s

# ------------------------------------------------------------
# 2. Imputación
# ------------------------------------------------------------

# 1. Categóricas: NaN → "Missing"
for col in categorical_cols:
    data[col] = data[col].cat.add_categories("Missing")
    data[col] = data[col].fillna("Missing")

# 2. Numéricas: imputación según porcentaje de missing
numeric_summary = summarize_numeric(data, numeric_cols)
low_missing = numeric_summary[numeric_summary["Missing %"].str.rstrip("%").astype(float) < 15].index
medium_missing = numeric_summary[
    (numeric_summary["Missing %"].str.rstrip("%").astype(float) >= 15) &
    (numeric_summary["Missing %"].str.rstrip("%").astype(float) <= 30)
].index
high_missing = numeric_summary[numeric_summary["Missing %"].str.rstrip("%").astype(float) > 30].index

print("Low (<15%):", np.array(low_missing))
print("Medium (15-30%):", np.array(medium_missing))
print("High to drop (>30%):", np.array(high_missing))

# a) Imputar mediana en low_missing
for col in low_missing:
    median_val = data[col].median()
    data[col] = data[col].fillna(median_val)

# b) Crear columna _missing antes de rellenar media
for col in medium_missing:
    missing_col = col + "_missing"
    data[missing_col] = data[col].isna().astype("int").astype("category")  # 1 = missing, 0 = original
    categorical_cols.append(missing_col)

# Ahora rellenar media en medium_missing
for col in medium_missing:
    mean_val = data[col].mean()
    data[col] = data[col].fillna(mean_val)

# c) Eliminar variables con >80% missing
data = data.drop(columns=high_missing)
numeric_cols = list(low_missing) + list(medium_missing)

print(f"Shape después de imputación", data.shape)
print("\n========== Resumen numérico post-imputación ==========")
display(summarize_numeric(data, numeric_cols))

print("\n=========== Resumen Variables Categóricas ===========")
display(summarize_categorical(data, categorical_cols))
```

La eliminación de las variables 'S_AD_KBRIG' 'D_AD_KBRIG' se sustenta, además del porcentaje de datos perdidos, sobre el análisis de correlación de variables hecha en el análisis exploratorio. Estas variables están fuertemente correlacionadas con 'S_AD_ORIT' y 'D_AD_ORIT', lo que hace que su eliminación no represente una perdida significativa de la información. La perdida de 'KFK_BLOOD' si representa una perdida de información, pero minima, puesto que esta variable tenia valor en un porcentaje muy mínimo de las instancias.

## 4. **One-hot encoding**

Las variables categóricas fueron transformadas mediante codificación One-Hot Encoding, lo cual permite representar cada categoría como una variable binaria independiente. Este procedimiento evita introducir un orden artificial entre categorías y es ampliamente utilizado en modelos de aprendizaje automático que requieren variables numéricas como entrada.

```python
# ============================================================
# One-Hot encoding
# ============================================================

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

# Obtener el dataframe de features
df_features = data.drop(columns=comp_cols + ["LET_IS"])
df_features[categorical_cols] = df_features[categorical_cols].astype(str)

# Instanciar el preprocesor de one-hot
preprocessor = ColumnTransformer(
    transformers=[
        ("onehot", OneHotEncoder(handle_unknown="ignore"), categorical_cols)
    ],
    remainder="passthrough"
)

# Aplicar transformación
X_numeric = preprocessor.fit_transform(df_features)

# Obtener nombres de columnas generadas
ohe_cols = preprocessor.named_transformers_["onehot"].get_feature_names_out(categorical_cols)
num_cols = [col for col in df_features.columns if col not in categorical_cols]
cols_final = list(ohe_cols) + num_cols

df_numeric = pd.DataFrame(X_numeric, columns=cols_final)
print("Shape después de One-Hot:", df_numeric.shape)
display(df_numeric.head())
```

Tras la codificación, el conjunto de datos pasó de 126 variables originales a 376 variables numéricas, debido a la expansión de las variables categóricas en múltiples indicadores binarios.

## 5. **Normalización**

La normalización de las variables numéricas se realizó considerando la forma de sus distribuciones observadas durante el análisis exploratorio. Las variables con distribución aproximadamente normal (edad, presión arterial y electrolitos) fueron estandarizadas mediante Z-score. En contraste, las variables con asimetría positiva (ALT_BLOOD, AST_BLOOD, L_BLOOD y ROE) fueron transformadas previamente mediante logaritmo natural para reducir la asimetría, y posteriormente estandarizadas mediante Z-score. Este procedimiento permite mejorar la estabilidad numérica de los modelos y evitar que variables con escalas muy diferentes dominen el proceso de aprendizaje.

```python
# ============================================================
# Normalización de variables numéricas según distribución
# ============================================================

from sklearn.preprocessing import StandardScaler
import numpy as np

# ------------------------------------------------------------
# 1. Definir grupos de variables según análisis exploratorio
# ------------------------------------------------------------

# Distribución aproximadamente normal
normal_cols = [
    "AGE",
    "S_AD_ORIT",
    "D_AD_ORIT",
    "NA_BLOOD",
    "K_BLOOD"
]


# Variables con asimetría positiva
skewed_cols = [
    "ALT_BLOOD",
    "AST_BLOOD",
    "L_BLOOD",
    "ROE"
]

cols_to_scale = normal_cols + skewed_cols

# ------------------------------------------------------------
# Boxplot ANTES de normalizar
# ------------------------------------------------------------
plt.figure(figsize=(5, 5))
data[cols_to_scale].boxplot()
plt.title("Boxplot ANTES de normalización")
plt.xticks(rotation=45)
plt.show()
print("\n\n")

# ------------------------------------------------------------
# 2. Reducir asimetría con transformación logarítmica
# ------------------------------------------------------------

df_numeric[skewed_cols] = np.log1p(df_numeric[skewed_cols])

# ------------------------------------------------------------
# 3. Aplicar estandarización Z-score
# ------------------------------------------------------------

scaler = StandardScaler()
cols_to_scale = normal_cols + skewed_cols
df_numeric[cols_to_scale] = scaler.fit_transform(
    df_numeric[cols_to_scale]
)

plt.figure(figsize=(5, 5))
df_numeric[cols_to_scale].boxplot()
plt.title("Boxplot DESPUÉS de normalización")
plt.xticks(rotation=45)
plt.show()
print("\n\n")

print("Normalización aplicada:")
print("- log-transform en variables asimétricas")
print("- Z-score en todas las variables numéricas\n")
display(df_numeric[cols_to_scale].head())
```

Para el problema de predicción de mortalidad, se utilizó una partición estratificada, garantizando que la proporción de las clases de la variable objetivo se mantuviera consistente en cada subconjunto. La división se realizó en tres etapas: primero se separó un 70% de los datos para entrenamiento y un 30% temporal, el cual posteriormente se dividió equitativamente para obtener los conjuntos de validación (15%) y prueba (15%).

Por otro lado, en el caso del problema multilabel de predicción de complicaciones, se utilizó un enfoque de estratificación iterativa (Iterative Stratification). Este método permite preservar de forma más adecuada la distribución conjunta de las diferentes etiquetas en los subconjuntos generados, lo cual resulta especialmente importante en problemas multilabel donde cada instancia puede pertenecer simultáneamente a múltiples clases. De esta forma, se generaron conjuntos de entrenamiento, validación y prueba manteniendo proporciones similares de cada complicación en los tres subconjuntos.

## 7. Selección de características

Con el fin de reducir la dimensionalidad del conjunto de datos y conservar únicamente las variables más relevantes para los modelos predictivos, se aplicó un proceso de selección de características basado en importancia de variables. Para ello, se utilizó un modelo de Random Forest, el cual permite estimar la contribución de cada variable al proceso de clasificación a partir de la reducción promedio de impureza generada por los árboles del bosque.

Dado que el estudio contempla dos problemas de predicción, se realizó la selección de características de manera diferenciada para cada uno. En el primer caso, correspondiente a la predicción de mortalidad hospitalaria (LET_IS), se entrenó un modelo Random Forest supervisado utilizando dicha variable como objetivo, obteniendo así un ranking de importancia de las características. Posteriormente, se seleccionó el subconjunto de variables con mayor relevancia para ser utilizado en el entrenamiento de los modelos.

```python
# ============================================================
# Importancia de características para problema multiclase
# ============================================================

from sklearn.ensemble import RandomForestClassifier
import pandas as pd
import numpy as np

# ------------------------------------------------------------
# 1. Definir variables predictoras y objetivo
# ------------------------------------------------------------

X = X_train
y = y_train

# ------------------------------------------------------------
# 2. Entrenar Random Forest simple
# ------------------------------------------------------------

rf = RandomForestClassifier(
    n_estimators=300,
    random_state=42,
    n_jobs=-1
)

rf.fit(X, y)

# ------------------------------------------------------------
# 3. Obtener importancia de variables
# ------------------------------------------------------------

importances = rf.feature_importances_

feat_importance = pd.DataFrame({
    "feature": X.columns,
    "importance": importances
})

feat_importance = feat_importance.sort_values(
    "importance", ascending=False
)

# mostrar top
display(feat_importance.head(20))
```

```python
# ============================================================
# Selección de top_k características
# ============================================================

top_k = 40

selected_features = feat_importance.head(top_k)["feature"]
X_selected = X[selected_features]

print("Shape original:", X.shape)
print("Shape reducido:", X_selected.shape)
```

En el segundo caso, correspondiente a la predicción de complicaciones clínicas, el problema se formula como una tarea multilabel, en la cual cada paciente puede presentar múltiples complicaciones simultáneamente. Para abordar esta situación, se entrenó un modelo Random Forest independiente para cada etiqueta de complicación y posteriormente se calculó la importancia promedio de cada característica a través de todos los modelos. Este enfoque permitió identificar las variables que, en promedio, resultan más informativas para la predicción conjunta de las distintas complicaciones.

```python
# ============================================================
# Importancia de características para problema multilabel
# ============================================================

from sklearn.ensemble import RandomForestClassifier
import pandas as pd
import numpy as np

X = X_train_multi
Y = Y_train_multi

feature_importances = []
for label in Y.columns:

    print(f"Entrenando modelo para: {label}")
    rf = RandomForestClassifier(
        n_estimators=300,
        random_state=42,
        n_jobs=-1
    )

    rf.fit(X, Y[label])
    feature_importances.append(rf.feature_importances_)

# ------------------------------------------------------------
# Promediar importancia entre todas las etiquetas
# ------------------------------------------------------------

feature_importances = np.array(feature_importances)
mean_importance = feature_importances.mean(axis=0)

feat_importance_multi = pd.DataFrame({
    "feature": X.columns,
    "importance": mean_importance
})

feat_importance_multi = feat_importance_multi.sort_values(
    "importance", ascending=False
)

print("\nTop variables para complicaciones:")
display(feat_importance_multi.head(20))
```

```python
# ============================================================
# Selección de top_k características
# ============================================================

top_k = 40

selected_features_multi = feat_importance_multi.head(top_k)["feature"]
X_selected_multi = X[selected_features_multi]

print("Shape original:", X.shape)
print("Shape reducido:", X_selected_multi.shape)
```
